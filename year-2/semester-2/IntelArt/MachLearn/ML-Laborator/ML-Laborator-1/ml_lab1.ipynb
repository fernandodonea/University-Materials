Se dau următoarele 9 imagini de dimensiuni 400x600. Valorile acestora au fost
salvate în fișierele “images/car_{idx}.npy”.

a. Citiți imaginile din aceste fișiere și salvați-le într-un np.array (va avea
dimensiunea 9x400x600).

```python
import numpy as np

images=np.zeros((9,400,600))
for i in range(0,9):
    img=np.load("images/car_"+str(i)+".npy")
    images[i]=img
print(images)



```
